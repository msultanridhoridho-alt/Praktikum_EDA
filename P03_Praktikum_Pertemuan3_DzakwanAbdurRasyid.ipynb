## Bagian D — Menandai data pribadi

| Kategori | Artinya |
|---|---|
| **pengenal langsung** | Sendirian sudah menunjuk ke satu orang |
| **quasi-identifier** | Sendirian aman, gabungannya menunjuk ke satu orang |
| **data spesifik** | Bocornya merugikan lebih berat (kesehatan, keuangan pribadi, data anak) |

```python
# TODO D1 — susun tabel penandaan Anda sendiri
PENANDAAN = {
    'nama_lengkap': ('pengenal langsung', 'hapus'),
    'nik': ('pengenal langsung', 'hapus'),
    'nomor_telepon': ('pengenal langsung', 'hapus'),
    'email': ('pengenal langsung', 'hash'),
    'id_peserta': ('pengenal langsung', 'hash'),
    'alamat_rumah': ('pengenal langsung', 'hapus'),
    'tanggal_lahir': ('quasi-identifier', 'generalisasi'),
    'kelurahan': ('quasi-identifier', 'hapus'),
    'kecamatan': ('quasi-identifier', 'generalisasi'),
    'jenis_kelamin': ('quasi-identifier', 'pertahankan'),
    'pendidikan_terakhir': ('quasi-identifier', 'pertahankan'),
    'penghasilan_pribadi_bulanan': ('data spesifik', 'pertahankan'),
    'jumlah_anak': ('data spesifik', 'pertahankan'),
    'status_pernikahan': ('data spesifik', 'pertahankan'),
}

# Bangun DataFrame 'tanda'
data_tanda = []
for col in df.columns:
    if col in PENANDAAN:
        kat, tind = PENANDAAN[col]
    else:
        kat, tind = ('bukan data pribadi', 'pertahankan')
    data_tanda.append({'nama_kolom': col, 'kategori': kat, 'tindakan': tind})

tanda = pd.DataFrame(data_tanda)

print("Jumlah kolom per kategori:")
print(tanda['kategori'].value_counts())
# Expected: pengenal langsung >= 5 kolom; data spesifik >= 3 kolom
```

## Bagian E — Menerapkan tiga tindakan

```python
# TODO E1 — HAPUS
kolom_dihapus = tanda[tanda['tindakan'] == 'hapus']['nama_kolom'].tolist()
aman = df.drop(columns=kolom_dihapus).copy()

print(f"Jumlah kolom yang dibuang : {len(kolom_dihapus)} -> {kolom_dihapus}")
print(f"Jumlah kolom yang tersisa  : {aman.shape[1]}")
```

```python
# TODO E2 — HASH
GARAM = 'ganti-dengan-kalimat-rahasia-tim'

def samarkan(nilai, garam=GARAM):
    if pd.isna(nilai):
        return np.nan
    gabungan = f"{garam}{nilai}"
    return hashlib.sha256(gabungan.encode('utf-8')).hexdigest()[:12]

kolom_hash = tanda[tanda['tindakan'] == 'hash']['nama_kolom'].tolist()
for col in kolom_hash:
    if col in aman.columns:
        aman[col] = aman[col].apply(lambda x: samarkan(x, GARAM))

print("Contoh hasil hash pada kolom tersisa:")
print(aman[kolom_hash].head(3))
```

## Bagian F — Memeriksa sisa risiko

Menghapus nama belum tentu cukup. Untuk setiap kombinasi quasi-identifier, hitung berapa orang yang punya kombinasi sama. Nilai terkecil disebut **k**. Target kelas ini: **k ≥ 5**.

```python
# TODO F1 — tulis fungsi cek_k(data, kolom)
def cek_k(data, kolom):
    grup = data.groupby(kolom).size()
    k_min = grup.min()
    grup_satu = (grup == 1).sum()
    total_grup = len(grup)

    return pd.Series({
        'k': k_min,
        'grup_berisi_1_orang': grup_satu,
        'total_grup': total_grup
    })
```

```python
# TODO F2 — ukur risiko SEBELUM penanganan
qi_sebelum = ['kelurahan', 'jenis_kelamin', 'pendidikan_terakhir', 'tanggal_lahir']
print(cek_k(df, qi_sebelum))
# Expected: k = 1 dan SELURUH 620 baris berada di grup berisi satu orang
```

```python
# TODO F4 — TANTANGAN: capai k >= 5
aman_k5 = aman.copy()

# Opsi Strategi:
# 1. Perlebar kelompok umur menjadi 3 kategori: '<35', '35-49', '50+'
bins_luas = [-1, 34, 49, 150]
labels_luas = ['<35', '35-49', '50+']
usia = (ACUAN - df['tanggal_lahir']).dt.days // 365.25
aman_k5['kelompok_umur'] = pd.cut(usia, bins=bins_luas, labels=labels_luas)

# 2. Keluarkan kolom 'pendidikan_terakhir' dari daftar quasi-identifier utama
qi_k5 = ['kecamatan', 'jenis_kelamin', 'kelompok_umur']

res_k5 = cek_k(aman_k5, qi_k5)
print(res_k5)
```
