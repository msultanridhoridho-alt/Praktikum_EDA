```python
# TODO E2 — HASH
# Tulis fungsi samarkan(nilai, garam) yang mengembalikan 12 karakter pertama
# dari sha256(garam + nilai). Kembalikan np.nan bila nilainya kosong.
# Terapkan pada kolom yang Anda beri tindakan 'hash'.
# Expected: nilai asli dan hasil hash berbeda, dan nilai asli yang sama menghasilkan hash yang sama

GARAM = 'ganti-dengan-kalimat-rahasia-tim'
```

```python
def samarkan(nilai, garam):
    if pd.isna(nilai):
        return np.nan
    gabungan = (str(garam) + str(nilai)).encode('utf-8')
    return hashlib.sha256(gabungan).hexdigest()[:12]

GARAM = 'ganti-dengan-kalimat-rahasia-tim'
kolom_hash = tanda[tanda['tindakan'] == 'hash']['nama_kolom'].tolist()

for col in kolom_hash:
    if col in aman.columns:
        aman[col] = aman[col].apply(lambda x: samarkan(x, GARAM))
```

```python
# TODO E3 — buktikan hash bersifat konsisten dan tidak terbaca
# a. Hash dua kali nilai yang sama, tunjukkan hasilnya identik
# b. Hash dua nilai berbeda, tunjukkan hasilnya berbeda

```

```python
val1 = samarkan('PLT-2025-1309', GARAM)
val2 = samarkan('PLT-2025-1309', GARAM)
print(f"Identik: {val1 == val2} -> {val1}")

val3 = samarkan('PLT-2025-1512', GARAM)
print(f"Berbeda: {val1 != val3} -> {val1} vs {val3}")
```

**Jawaban Anda:** kalau `GARAM` ikut ter-*commit* ke GitHub, apa yang masih terlindungi dan apa yang tidak?

Jika GARAM ikut ter-commit, hash menjadi sangat rentan terhadap teknik brute-force atau serangan kamus (rainbow table). Data pribadi yang memiliki pola logis atau tetap (seperti kombinasi tanggal lahir, struktur ID, atau nama kelurahan) dapat dijebol, dan satu-satunya perlindungan tersisa hanyalah pada data yang kebetulan memiliki entropi dan tingkat keacakan yang sangat tinggi/panjang.

```python
# TODO E4 — GENERALISASI
# a. Hitung usia peserta pada tanggal acuan 1 September 2025
# b. Buat kolom 'kelompok_umur' dengan batas: <25, 25-34, 35-44, 45-54, 55+
# c. Buang kolom tanggal_lahir dan kelurahan dari 'aman'
# d. Rapikan kolom kecamatan yang sebagian tertulis KAPITAL
# Expected: kelompok terbanyak adalah 35-44; kecamatan punya tepat 10 nilai unik

ACUAN = pd.Timestamp('2025-09-01')
```

```python
tanggal_lahir = pd.to_datetime(df['tanggal_lahir'])

usia_tahun = (
    ACUAN.year
    - tanggal_lahir.dt.year
    - (
        (ACUAN.month < tanggal_lahir.dt.month)
        | (
            (ACUAN.month == tanggal_lahir.dt.month)
            & (ACUAN.day < tanggal_lahir.dt.day)
        )
    ).astype(int)
)

bins = [-1, 24, 34, 44, 54, 150]
labels = ['<25', '25-34', '35-44', '45-54', '55+']

aman['kelompok_umur'] = pd.cut(
    usia_tahun,
    bins=bins,
    labels=labels
)

aman = aman.drop(
    columns=['tanggal_lahir', 'kelurahan'],
    errors='ignore'
)

aman['kecamatan'] = aman['kecamatan'].str.title()

print(aman['kelompok_umur'].value_counts())
print("Jumlah kecamatan unik:", aman['kecamatan'].nunique())
```

## Bagian F — Memeriksa sisa risiko

Menghapus nama belum tentu cukup. Untuk setiap kombinasi quasi-identifier, hitung berapa orang yang punya kombinasi sama. Nilai terkecil disebut **k**. Target kelas ini: **k ≥ 5**.

```python
# TODO F1 — tulis fungsi cek_k(data, kolom)
# Kembalikan Series berisi: k (grup terkecil), grup berisi 1 orang, total grup

```

```python
def cek_k(data, kolom):
    grup = data.groupby(kolom, observed=True).size()
    return pd.Series({
        'k': grup.min(),
        'grup_1_orang': (grup == 1).sum(),
        'total_grup': len(grup)
    })
```

```python
# TODO F2 — ukur risiko SEBELUM penanganan
# Gunakan data asli df dengan kolom:
#   ['kelurahan', 'jenis_kelamin', 'pendidikan_terakhir', 'tanggal_lahir']
# Expected: k = 1 dan SELURUH 620 baris berada di grup berisi satu orang
```

```python
print(cek_k(df, ['kelurahan', 'jenis_kelamin', 'pendidikan_terakhir', 'tanggal_lahir']))
```

**Jawaban Anda:** keempat kolom itu tidak memuat nama maupun nomor telepon. Jelaskan dalam dua kalimat mengapa hasilnya tetap berbahaya.

Meskipun pengenal mutlak (nama/nomor) nihil, persinggungan antara tanggal lahir spesifik, jenjang pendidikan, jenis kelamin, dan letak demografi (kelurahan) menghasilkan fingerprint unik. Aktor luar cukup memiliki akses ke satu pos data terpisah (misalnya, kartu keluarga warga kelurahan setempat, profil Linkedin warga, atau biodata tetangga terdekat) dan akan mampu merekonstruksi serta menunjuk baris mana milik siapa secara presisi.

```python

# G4 — Penanganan dataset Bakery dan pemeriksaan k-anonymity

import pandas as pd
import os

# Membuat folder penyimpanan
os.makedirs('data/processed', exist_ok=True)

# Membaca dataset
file_path = 'Bakery sales (2).csv'
df_tim = pd.read_csv('/content/drive/MyDrive/Eda/Tugas Praktikum P04/Bakery sales (2).csv')

# Membuat salinan data
aman_tim = df_tim.copy()

# Menghapus kolom indeks yang tidak diperlukan
aman_tim = aman_tim.drop(
    columns=['Unnamed: 0'],
    errors='ignore'
)

# Membersihkan format tanggal
aman_tim['date'] = pd.to_datetime(
    aman_tim['date'],
    errors='coerce'
)

# Menghapus baris yang tanggalnya tidak valid
aman_tim = aman_tim.dropna(subset=['date'])

# Generalisasi waktu:
# Informasi jam dihapus karena kombinasi tanggal dan jam
# masih menghasilkan kelompok dengan k = 1.
aman_tim = aman_tim.drop(
    columns=['time'],
    errors='ignore'
)

# Menghitung jumlah transaksi unik pada setiap tanggal
transaksi = aman_tim.drop_duplicates(
    subset=['ticket_number']
)

grup = transaksi.groupby('date').size()

# Menghitung nilai k-anonymity
hasil_k = pd.Series({
    'k': grup.min(),
    'grup_berisi_1_transaksi': (grup == 1).sum(),
    'grup_kurang_dari_5': ((grup > 0) & (grup < 5)).sum(),
    'total_grup': len(grup)
})

print("Hasil pemeriksaan k-anonymity:")
print(hasil_k)

if hasil_k['k'] >= 5:
    print("\nData memenuhi target k-anonymity minimal 5.")
else:
    print("\nData belum memenuhi target k-anonymity minimal 5.")

# Menghapus nomor transaksi asli dari data yang dibagikan
aman_tim = aman_tim.drop(
    columns=['ticket_number'],
    errors='ignore'
)

# Menyimpan data hasil penanganan
path_simpan = 'data/processed/bakery_sales_processed.csv'

aman_tim.to_csv(path_simpan, index=False)

print("\nData hasil penanganan berhasil disimpan:")
print(path_simpan)
```
