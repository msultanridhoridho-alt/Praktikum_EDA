## Bagian E — Menerapkan tiga tindakan

```python
# TODO E1 — HAPUS
kolom_dihapus = tanda[tanda['tindakan'] == 'hapus']['nama_kolom'].tolist()
aman = df.drop(columns=kolom_dihapus).copy()

print(f"Jumlah kolom yang dibuang : {len(kolom_dihapus)} -> {kolom_dihapus}")
print(f"Jumlah kolom yang tersisa  : {aman.shape[1]}")
```

```python
# TODO E2 — HASH
GARAM = 'ganti-dengan-kalimat-rahasia-tim'

def samarkan(nilai, garam=GARAM):
    if pd.isna(nilai):
        return np.nan
    gabungan = f"{garam}{nilai}"
    return hashlib.sha256(gabungan.encode('utf-8')).hexdigest()[:12]

kolom_hash = tanda[tanda['tindakan'] == 'hash']['nama_kolom'].tolist()
for col in kolom_hash:
    if col in aman.columns:
        aman[col] = aman[col].apply(lambda x: samarkan(x, GARAM))

print("Contoh hasil hash pada kolom tersisa:")
print(aman[kolom_hash].head(3))
```

```python
# TODO E3 — buktikan hash bersifat konsisten dan tidak terbaca
# a. Hash dua kali nilai yang sama
val1 = samarkan("peserta_001", GARAM)
val2 = samarkan("peserta_001", GARAM)
print(f"Hash 1: {val1}\nHash 2: {val2}\nSama? {val1 == val2}")

# b. Hash dua nilai berbeda
val3 = samarkan("peserta_002", GARAM)
print(f"Hash 3: {val3}\nBerbeda dari Hash 1? {val1 != val3}")
```

**Jawaban Anda:** kalau `GARAM` ikut ter-*commit* ke GitHub, apa yang masih terlindungi dan apa yang tidak?

_(Jika GARAM ikut ter-commit ke repositori publik GitHub, perlindungan berbasis salt hilang dari ancaman rainbow table atau dictionary attack. Seseorang yang memiliki daftar data mentah (seperti daftar email atau ID) dapat menghitung nilai hash-nya sendiri dengan salt tersebut dan mencocokkannya untuk mengidentifikasi identitas pemilik data (re-identification). Hal yang masih terlindungi hanyalah pembalikan hash secara langsung tanpa tabel pencarian (reverse lookup satu arah fungsi SHA-256).)_

```python
# TODO E4 — GENERALISASI
ACUAN = pd.Timestamp('2025-09-01')

# a. Hitung usia peserta pada tanggal acuan 1 September 2025
usia = (ACUAN - df['tanggal_lahir']).dt.days // 365.25

# b. Buat kolom 'kelompok_umur' dengan batas: <25, 25-34, 35-44, 45-54, 55+
bins = [-1, 24, 34, 44, 54, 150]
labels = ['<25', '25-34', '35-44', '45-54', '55+']
aman['kelompok_umur'] = pd.cut(usia, bins=bins, labels=labels)

# c. Buang kolom tanggal_lahir dan kelurahan dari 'aman'
kolom_dibuang_e4 = [col for col in ['tanggal_lahir', 'kelurahan'] if col in aman.columns]
aman = aman.drop(columns=kolom_dibuang_e4)

# d. Rapikan kolom kecamatan yang sebagian tertulis KAPITAL
aman['kecamatan'] = aman['kecamatan'].str.title()

print("Distribusi kelompok umur:")
print(aman['kelompok_umur'].value_counts())
print(f"\nJumlah nilai unik kecamatan: {aman['kecamatan'].nunique()}")
# Expected: kelompok terbanyak adalah 35-44; kecamatan punya tepat 10 nilai unik
```

## Bagian F — Memeriksa sisa risiko

Menghapus nama belum tentu cukup. Untuk setiap kombinasi quasi-identifier, hitung berapa orang yang punya kombinasi sama. Nilai terkecil disebut **k**. Target kelas ini: **k ≥ 5**.

```python
# TODO F1 — tulis fungsi cek_k(data, kolom)
def cek_k(data, kolom):
    grup = data.groupby(kolom).size()
    k_min = grup.min()
    grup_satu = (grup == 1).sum()
    total_grup = len(grup)

    return pd.Series({
        'k': k_min,
        'grup_berisi_1_orang': grup_satu,
        'total_grup': total_grup
    })
```

```python
# TODO F2 — ukur risiko SEBELUM penanganan
qi_sebelum = ['kelurahan', 'jenis_kelamin', 'pendidikan_terakhir', 'tanggal_lahir']
print(cek_k(df, qi_sebelum))
# Expected: k = 1 dan SELURUH 620 baris berada di grup berisi satu orang
```

**Jawaban Anda:** keempat kolom itu tidak memuat nama maupun nomor telepon. Jelaskan dalam dua kalimat mengapa hasilnya tetap berbahaya.

_(Meskipun keempat kolom tersebut tidak memuat nama atau nomor telepon, kombinasi spesifik dari tanggal lahir persis, kelurahan, jenis kelamin, dan pendidikan terakhir sangat bersifat unik bagi tiap individu. Penyerang yang mengetahui informasi umum target dari media sosial atau interaksi langsung dapat dengan mudah mencocokkan kombinasi ini untuk mengidentifikasi baris data milik orang tersebut secara pasti.)_

```python
# TODO F3 — ukur risiko SESUDAH generalisasi
qi_sesudah = ['kecamatan', 'jenis_kelamin', 'pendidikan_terakhir', 'kelompok_umur']
print(cek_k(aman, qi_sesudah))
# Expected: jumlah grup berisi satu orang turun banyak, tetapi k masih 1
```

```python
# TODO F4 — TANTANGAN: capai k >= 5
aman_k5 = aman.copy()

# Opsi Strategi:
# 1. Perlebar kelompok umur menjadi 3 kategori: '<35', '35-49', '50+'
bins_luas = [-1, 34, 49, 150]
labels_luas = ['<35', '35-49', '50+']
usia = (ACUAN - df['tanggal_lahir']).dt.days // 365.25
aman_k5['kelompok_umur'] = pd.cut(usia, bins=bins_luas, labels=labels_luas)

# 2. Keluarkan kolom 'pendidikan_terakhir' dari daftar quasi-identifier utama
qi_k5 = ['kecamatan', 'jenis_kelamin', 'kelompok_umur']

res_k5 = cek_k(aman_k5, qi_k5)
print(res_k5)
```

**Jawaban Anda:** tulis kombinasi yang Anda pilih, nilai k yang dicapai, dan informasi apa yang Anda korbankan untuk mencapainya.

_(Kombinasi yang dipilih:Menggeneralisasikan kelompok_umur menjadi rentang yang lebih lebar (<35, 35-49, 50+) dan mengeluarkan pendidikan_terakhir dari tuple pemetaan quasi-identifier publik (['kecamatan', 'jenis_kelamin', 'kelompok_umur']).Nilai k yang dicapai: $k \ge 5$ (seluruh grup memiliki minimal 5 anggota).Informasi yang dikorbankan: Kedetailan granulitas usia dan hubungan silang tingkat pendidikan per kecamatan untuk memprioritaskan privasi subjek data secara penuh.)_

```python
# TODO F5 — sel kecil pada tabel agregat
# a. Tabel silang jumlah peserta menurut kecamatan x kategori_usaha
tabel_silang = pd.crosstab(aman['kecamatan'], aman['kategori_usaha'])

# b. Hitung berapa sel yang terisi tetapi berisi kurang dari 5 peserta
sel_kecil = ((tabel_silang > 0) & (tabel_silang < 5)).sum().sum()

print(tabel_silang)
print(f"\nJumlah sel terisi bernilai 1-4: {sel_kecil}")
# Expected: ada sel bernilai 1-4; sel seperti ini tidak boleh ditampilkan apa adanya di laporan
```
